```python
import pandas as pd
```

### Concatenate datasets

```python
# load all datasets
dataset1 = pd.read_csv('pilot_sample_run1_mscreened.csv')
dataset2 = pd.read_csv('pilot_sample_run2_mscreened.csv')
dataset3 = pd.read_csv('screening_results_glm.csv')
```

```python
# concat the datasets
concat_dataset = pd.concat([dataset1, dataset2, dataset3], ignore_index=True)

# check for duplicates
duplicates = concat_dataset[concat_dataset.duplicated(subset='source_id', keep=False)]
duplicates
```

```python
# save to csv
concat_dataset.to_csv('concatenated_datasets.csv', index=False)
```

## Correction: Remove all rows in dataset1 from the concatenated dataset (were reviewed using different criteria)

```python
# load the concatenated dataset
concat_dataset = pd.read_csv('concatenated_datasets.csv')
```

```python
# remove all rows in dataset1 from concat_dataset
concat_dataset = concat_dataset[~concat_dataset['source_id'].isin(dataset1['source_id'])]
```

```python
concat_dataset.to_csv('concatenated_datasets.csv', index=False)
```

## Add the remaining subset of 100 papers that were re-screened using the GLM model and the new criteria (v9)

```python
# load the concatenated dataset
concat_dataset = pd.read_csv('concatenated_datasets.csv')
# load the subset
subset = pd.read_csv('screening_results_glm_subset1.csv')
```

```python
#concatenate the subset to the concatenated dataset
concat_dataset = pd.concat([concat_dataset, subset], ignore_index=True)
```

```python
# save to csv
concat_dataset.to_csv('concatenated_datasets.csv', index=False)
```

## Extract 25% randomly sampled exclusions for manual check

```python
# load dataset
dataset1 = pd.read_csv("screening_results_glm.csv")
dataset2 = pd.read_csv("screening_results_glm_subset1.csv")

#concat the datasets
dataset3 = pd.concat([dataset1, dataset2], ignore_index=True)
dataset3_exclusions = dataset3[dataset3['decision'] == 'Exclude']

# sample random 25%
seed = 42
dataset3_exclusions_sample_25 = dataset3_exclusions.sample(frac=0.25, random_state=seed)

# export
dataset3_exclusions.to_csv('screening_results_glm_exclusions.csv', index=False)

dataset3_exclusions_sample_25.to_csv('screening_results_glm_exclusions_sample_25.csv', index=False)

```
