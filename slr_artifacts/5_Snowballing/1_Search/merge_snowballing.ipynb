```python
import pandas as pd
import os
```

```python
# iterate through all the csv files in the directory and merge them into a single dataframe
directory = "snowball_out"
merged_df = pd.DataFrame()

for file in os.listdir(directory):
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(directory, file))
        if 'merged_df' in locals():
            merged_df = pd.concat([merged_df, df], ignore_index=True)
        else:
            merged_df = df

# save te file
merged_df.to_csv("snowball_out.csv")
merged_df
```

```python
# check against existing dataset and remove duplicates
existing_dataset = pd.read_csv("fulltext-screened-dataset.csv")

merged_df = merged_df[~merged_df['source_id'].isin(existing_dataset['source_id'])]
duplicates_df = merged_df[merged_df['source_id'].isin(existing_dataset['source_id'])]
print(duplicates_df)

# save the final merged dataset
merged_df.to_csv("snowball_out.csv")
merged_df
```
